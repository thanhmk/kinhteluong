# **LỘ TRÌNH KHÓA HỌC KINH TẾ LƯỢNG TÀI CHÍNH**
*7 module · 28 bài · khoảng 29 giờ học (60 phút mỗi bài, riêng Module 7 – Bài 4 là 90 phút), chưa tính bài tập tự luyện*

## **1. Mục tiêu khóa học**
Sau khóa học, bạn có thể:
1. Ước lượng, chẩn đoán và diễn giải mô hình **hồi quy** (OLS, WLS, vững, phạt, phi tham số) cho dữ liệu tài chính.
2. Mô tả **phân phối lợi suất** (đuôi dày, lệch) và **cấu trúc phụ thuộc** (tương quan, copula).
3. Dựng mô hình **chuỗi thời gian** đơn biến (ARMA, ARIMA) và mô hình **biến động** (ARCH, GARCH, Bayes, không gian trạng thái).
4. Phân tích **đa biến và dài hạn**: nghiệm đơn vị, VAR, đồng tích hợp, ECM, VECM.
5. Dùng **nhân tố, mạng, nhân quả Granger** để hiểu cấu trúc kết nối, và nắm các khái niệm mở đầu của **học máy** (LDA, phân cụm).

## **2. Điều kiện đầu vào**
- Xác suất – thống kê cơ bản (kỳ vọng, phương sai, kiểm định giả thuyết); đại số tuyến tính cơ bản (ma trận, giá trị riêng).
- Python cơ bản (pandas, numpy, matplotlib).

## **3. Cách học mỗi bài**
Mỗi sổ tay theo cùng một nhịp: **Mục tiêu → Lý thuyết (công thức, trực giác) → Thực hành (code chạy được trên dữ liệu đi kèm) → Nhận xét kết quả → Tóm tắt → Bài tập tự luyện.** Gợi ý: đọc lý thuyết trước, *tự đoán* kết quả, rồi chạy code và đối chiếu; làm ít nhất một bài tập cuối bài.

## **5. Bản đồ kiến thức (thứ tự phụ thuộc)**
```
M1 Hồi quy ─► M2 Hồi quy nâng cao ──┐
     │                              ├─► M7-B1 Nhân tố / M7-B2 Mạng (dùng LASSO, PCA)
     └─► M1-B3 PCA ─────────────────┘
M3 Phân phối & Copula ─► M5 GARCH (đuôi dày, Student-t)
M4 Chuỗi thời gian ─► M5 ARCH/GARCH ─► M5-B3 Bayes, M5-B4 Kalman
M4 ─► M6-B1 Nghiệm đơn vị ─► M6-B2 VAR ─► M6-B3 ECM ─► M6-B4 VECM ─► M7-B3 Granger
```
Có thể học **M1–M2** và **M3** song song; **M4 phải đi trước M5 và M6**.

## **8. Ghi chú về bộ sổ tay**
- Mọi sổ tay là **bản tiếng Việt viết lại**; code chạy lại được trên dữ liệu đi kèm. Số liệu đã được đối chiếu với kết quả khi chạy.
- Một số điểm khác bản gốc: (i) các đoạn code vốn viết bằng ngôn ngữ R hoặc không chạy được trên phiên bản thư viện hiện hành được viết lại bằng Python (ví dụ copula Clayton/Gumbel ở Module 3 – Bài 4; chẩn đoán GARCH ở Module 5 – Bài 2); (ii) **Module 5 – Bài 4** và **Module 7 – Bài 4** vốn chỉ có lý thuyết, đã **bổ sung phần thực hành** (bộ lọc Kalman tự viết; LDA/phân cụm trên dữ liệu mô phỏng); (iii) Module 6 – Bài 2 chuyển dự báo về mức bằng cách cộng dồn log-sai phân rồi lấy mũ; Module 6 – Bài 3 kiểm định Durbin-Watson trên phần dư (không phải giá trị khớp).
- **Module 7 – Bài 2 và 3** dùng dữ liệu thị trường tải về, nên số liệu có thể khác nhẹ so với mô tả tùy ngày tải; sổ tay lưu sẵn mô tả theo hướng dẫn cách đọc, không chốt con số.
- Một số kết quả (ví dụ Module 5 – Bài 3) có phần ngẫu nhiên của MCMC; đã cố định `random_seed` nhưng con số có thể lệch nhẹ giữa các phiên bản PyMC.
